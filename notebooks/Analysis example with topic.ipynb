```python
import rouge 

# candidates = ['menteri luar negeri amerika serikat , john kerry , mengatakan rencana untuk mendukung rencana untuk membahas masalah di suriah dan suriah .']
candidates = ['para menteri luar negeri amerika serikat dan rusia di jenewa untuk membahas rencana rusia untuk membahas krisis senjata kimia untuk mengakhiri perang suriah . ']
golds = ['menteri luar negeri amerika serikat dan rusia telah mulai pembicaraan penting di jenewa mengenai rencana untuk pengendalian senjata kimia suriah oleh pihak internasional .']

for aggregator in ['Individual']:

    evaluator = rouge.Rouge(metrics=['rouge-n', 'rouge-l'],
                           max_n=3,
                           limit_length=False,
                           alpha=0.5, # Default F1_score
                           apply_avg=False,
                           apply_best=False,
                           weight_factor=1.0,
                           stemming=True)
    
    scores = evaluator.get_scores(candidates, golds)
    
    results_list = [0 for i in range(len(candidates))]
    for metric, results in sorted(scores.items(), key=lambda x: x[0]):
        
        for cand_id, results_per_ref in enumerate(results):
            if metric == 'rouge-1':
                results_list[cand_id] = [results_per_ref['f'][0]]
            else:
                results_list[cand_id].append(results_per_ref['f'][0])

r1 = []
r2 = []
r3 = []
rl = []
for res in results_list:
    r1.append(res[0])
    r2.append(res[1])
    r3.append(res[2])
    rl.append(res[3])

print("ROUGE Scores")
print(f"ROUGE-1: {sum(r1) * 100}")
print(f"ROUGE-2: {sum(r2) * 100}")
print(f"ROUGE-L: {sum(rl) * 100}")
```

```python
from sentence_transformers import SentenceTransformer
from bertopic import BERTopic
```

```python
embedding_model = SentenceTransformer("denaya/indoSBERT-large")
```

```python
loaded_model = BERTopic.load("../topic_models/model_5", embedding_model=embedding_model)
```

```python
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

# Remove stop words
def remove_stop_words(doc):
    factory = StopWordRemoverFactory()
    stopword = factory.create_stop_word_remover()
    res = stopword.remove(doc)
    return res
```

```python
# `topic_distr` contains the distribution of topics in each document
topic_distr, _ = loaded_model.approximate_distribution(proc_test_docs)
```
