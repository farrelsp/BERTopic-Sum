```python
import os
import json

def analyze_tokens(path):
    files = []
    for file in os.listdir(path):
        if '.json' in file:
            files.append(file)
    
    files = sorted(files)
    
    min_tokens = 9999999
    max_tokens = 0
    tokens_length = []
    for file in files:
        filename = path + file
        with open(filename) as f:
            batch_data = json.load(f)
            for data in batch_data:
                length = len(data['tgt'])
                tokens_length.append(length)
                if length > max_tokens:
                    max_tokens = length
                if length < min_tokens:
                    min_tokens = length

    print("Statistics")
    print("Min. tokens:", min_tokens)
    print("Max tokens:", max_tokens)
    print("Average tokens:", sum(tokens_length)/len(tokens_length))
```

```python
path = "../bert_data/en/"
analyze_tokens(path)
```

```python
path = "../bert_data/id-p1/"
analyze_tokens(path)
```
